```python
filtered_app_df = app_df[
    (app_df['Reviews'] > 500) &
    (~app_df['App'].str.contains('S', case=False, na=False)) &
    (~app_df['App'].str.startswith(('X', 'Y', 'Z'), na=False)) &
    (app_df['Category'].str.startswith(('B', 'C', 'E'), na=False))
].copy()
```

```python
filtered_app_df['Category_Display'] = filtered_app_df['Category'].replace({
    'BEAUTY': 'सौंदर्य',
    'BUSINESS': 'வணிகம்',
    'DATING': 'Partnersuche'
})
```

```python
monthly_installs = (
    filtered_app_df
    .groupby(['Month', 'Category_Display'], as_index=False)['Installs']
    .sum()
)
```

```python
monthly_installs['MoM_Growth'] = (
    monthly_installs
    .groupby('Category_Display')['Installs']    .pct_change()
)
```

```python
filtered_stacked_df = app_df[
    (app_df['Rating'] >= 4.2) &
    (app_df['Reviews'] > 1000) &
    (app_df['Size'] >= 20) &
    (app_df['Size'] <= 80) &
    (~app_df['App'].str.contains(r'\d', regex=True, na=False)) &
    (app_df['Category'].str.startswith(('T', 'P'), na=False))
].copy()
```

```python
filtered_stacked_df['Category_Display'] = (
    filtered_stacked_df['Category'].replace({
        'TRAVEL & LOCAL': 'Voyage et local',
        'PRODUCTIVITY': 'Productividad',
        'PHOTOGRAPHY': '写真'
    })
)
```

```python
filtered_stacked_df['Category_Display'] = filtered_stacked_df['Category_Display'].replace({
    'TRAVEL_AND_LOCAL': 'Voyage et local'
})
```

```python
filtered_stacked_df['Last Updated'] = pd.to_datetime(
    filtered_stacked_df['Last Updated']
)

filtered_stacked_df['Month'] = (
    filtered_stacked_df['Last Updated']
    .dt.to_period('M')
    .astype(str)
)
```

```python
stacked_monthly = (
    filtered_stacked_df
    .groupby(['Month', 'Category_Display'], as_index=False)['Installs']
    .sum()
)
```

```python
stacked_monthly = stacked_monthly.sort_values(
    ['Category_Display', 'Month']
)

stacked_monthly['MoM_Growth'] = (
    stacked_monthly
    .groupby('Category_Display')['Installs']
    .pct_change()
)
```

```python
stacked_monthly['Growth_Flag'] = (
    stacked_monthly['MoM_Growth'] > 0.25
)
```

```python
growth_months = (
    stacked_monthly
    .groupby('Month')['Growth_Flag']
    .any()
    .reset_index(name='Growth_Month')
)
```

```python
stacked_monthly = stacked_monthly.merge(
    growth_months,
    on='Month',
    how='left'
)
```

```python
stacked_monthly['Cumulative_Installs'] = (
    stacked_monthly
    .groupby('Category_Display')['Installs']
    .cumsum()
)
```

```python
stacked_monthly.to_csv('stacked_monthly.csv',index=False)
```
